## Solución del taller

**1. Menor de dos pares:** escribir una función que devuelva el menor de dos números si ambos son pares, o el mayor si alguno es impar.
```
lesser_of_two_evens(2,4) --> 2
lesser_of_two_evens(2,5) --> 5
```

````python
def lesser_of_two_evens(a, b):
    if a % 2 == 0 and b % 2 == 0:
        return min(a, b)
    else:
        return max(a, b)

lesser_of_two_evens(2, 4), lesser_of_two_evens(2, 5)
````

**Explicación:** Reviso primero si los dos números son pares con `%`; si es así devuelvo el menor con `min`, y si no, el mayor con `max`, tal como lo pide el enunciado.

**2. Galletas de animales:** dada una cadena de dos palabras, devolver `True` si ambas empiezan con la misma letra.
```
animal_crackers('Ciudad Calor') --> True
animal_crackers('Lugar Frio') --> False
```

````python
def animal_crackers(text):
    palabras = text.split()
    return palabras[0][0].lower() == palabras[1][0].lower()

animal_crackers('Ciudad Calor'), animal_crackers('Lugar Frio')
````

**Explicación:** Con `split()` separo la cadena en dos palabras y comparo su primer carácter (`[0]`) en minúscula, para que la comparación no dependa de mayúsculas o minúsculas.

**3. Hace veinte:** dados dos enteros, devolver `True` si su suma es 20 o si alguno de los dos es 20.
```
makes_twenty(20,10) --> True
makes_twenty(12,8) --> True
makes_twenty(2,3) --> False
```

````python
def makes_twenty(n1, n2):
    return (n1 + n2 == 20) or (n1 == 20) or (n2 == 20)

makes_twenty(20, 10), makes_twenty(12, 8), makes_twenty(2, 3)
````

**Explicación:** Uso `or` para juntar las tres condiciones del enunciado (la suma es 20, o el primero es 20, o el segundo es 20) en una sola línea.

**4. Mayúsculas:** hacer mayúscula la primera y la cuarta letra de una cadena.
```
old_macdonald('macdonald') --> MacDonald
```

````python
def old_macdonald(name):
    return name[:3].capitalize() + name[3].upper() + name[4:]

old_macdonald('macdonald')
````

**Explicación:** Parto la cadena en tres pedazos con slicing: los primeros tres caracteres (`capitalize()` para que solo el primero quede en mayúscula), el cuarto carácter por separado (`upper()`), y el resto tal cual.

**5. Reversa (Master Yoda):** dada una oración, devolverla con las palabras en orden inverso.
```
master_yoda('I am home') --> 'home am I'
master_yoda('We are ready') --> 'ready are We'
```

````python
def master_yoda(text):
    palabras = text.split()
    return " ".join(reversed(palabras))

master_yoda('I am home'), master_yoda('We are ready')
````

**Explicación:** `split()` separa la oración en una lista de palabras, `reversed()` las recorre en orden inverso, y `" ".join(...)` las vuelve a unir en una sola cadena separadas por espacio.

**6. Problema 33:** dada una lista de enteros, devolver `True` si hay un 3 seguido inmediatamente por otro 3.
```
has_33([1, 3, 3]) --> True
has_33([1, 3, 1, 3]) --> False
has_33([3, 1, 3]) --> False
```

````python
def has_33(nums):
    for i in range(len(nums) - 1):
        if nums[i] == 3 and nums[i + 1] == 3:
            return True
    return False

has_33([1, 3, 3]), has_33([1, 3, 1, 3]), has_33([3, 1, 3])
````

**Explicación:** Recorro la lista comparando cada elemento con el siguiente (`nums[i]` y `nums[i+1]`); en cuanto encuentro dos 3 seguidos devuelvo `True` de inmediato, y si termino el ciclo sin encontrarlos devuelvo `False`.

**7. Replicador:** dada una cadena, devolver una nueva donde cada carácter aparece tres veces seguidas.
```
paper_doll('Hello') --> 'HHHeeellllllooo'
```

````python
def paper_doll(text):
    resultado = ""
    for caracter in text:
        resultado += caracter * 3
    return resultado

paper_doll('Hello')
````

**Explicación:** Recorro cada carácter de la cadena y lo repito con `caracter * 3` (el mismo operador de repetición que se vio en el cuaderno de cadenas), acumulando el resultado en una cadena nueva.

**8. Blackjack:** dados tres enteros entre 1 y 11, devolver su suma si es menor o igual a 21; si la excede y hay un 11 entre ellos, restar 10; si aun así excede 21, devolver `'BUST'`.
```
blackjack(5,6,7) --> 18
blackjack(9,9,9) --> 'BUST'
blackjack(9,9,11) --> 19
```

````python
def blackjack(a, b, c):
    total = a + b + c
    if total <= 21:
        return total
    if 11 in (a, b, c):
        total -= 10
    if total > 21:
        return 'BUST'
    return total

blackjack(5, 6, 7), blackjack(9, 9, 9), blackjack(9, 9, 11)
````

**Explicación:** Primero calculo la suma total; si ya cabe en 21 la devuelvo directo. Si no, reviso con el operador `in` si alguno de los tres números es 11 para restarle 10; y solo si después de ese ajuste sigue pasándose de 21, devuelvo `'BUST'`.
