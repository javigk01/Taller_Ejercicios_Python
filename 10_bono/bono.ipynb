#### MENOR DE DOS PARES: Escriba una función que devuelva el menor de dos números dados *si* ambos números son pares, pero devuelva el mayor si uno o ambos números son impares
    lesser_of_two_evens(2,4) --> 2
    lesser_of_two_evens(2,5) --> 5

```python
def lesser_of_two_evens(a,b):
    if a % 2 == 0 and b % 2 == 0:
        return min(a, b)
    else:
        return max(a, b)
```

**Mi explicación (lesser_of_two_evens):** Si ambos numeros son pares devuelvo el menor con min(); si al menos uno es impar devuelvo el mayor con max(). La condicion a%2==0 and b%2==0 es justo 'ambos pares'.

```python
# Revisión
lesser_of_two_evens(2,4)
```

```python
# Revisión
lesser_of_two_evens(2,5)
```

#### GALLETAS DE ANIMALES: Escribir una función que toma una cadena de dos palabras y devuelve True si ambas palabras comienzan con la misma letra
    animal_crackers('Ciudad Calor') --> True
    animal_crackers('Lugar Frio') --> False

```python
def animal_crackers(text):
    palabras = text.split()
    return palabras[0][0].lower() == palabras[1][0].lower()
```

**Mi explicación (animal_crackers):** Separo el texto en dos palabras con split() y comparo la primera letra de cada una (en minuscula para que 'Ciudad'/'Calor' cuenten como la misma letra sin importar mayusculas).

```python
# Revisión
animal_crackers('Ciudad Calor')
```

```python
# Revisión
animal_crackers('Lugar Frio')
```

#### HACE VEINTE: dados dos enteros, devuelve True si la suma de los enteros es 20 o si uno de los enteros es 20. De lo contrario, devuelve False

    makes_twenty(20,10) --> True
    makes_twenty(12,8) --> True
    makes_twenty(2,3) --> False

```python
def makes_twenty(n1,n2):
    return (n1 + n2 == 20) or (n1 == 20) or (n2 == 20)
```

**Mi explicación (makes_twenty):** Uso or para cubrir los tres casos que pide el enunciado en una sola linea: que la suma de los dos sea 20, o que cualquiera de los dos ya sea 20 por si solo.

```python
# Revisión
makes_twenty(20,10)
```

```python
# Revisión
makes_twenty(2,3)
```

```python
makes_twenty(12,8)
```

#### MAYÚSCULAS: Escribir una función que haga mayúscula la primera y cuarta letra de una cadena de texto
     
    old_macdonald('macdonald') --> MacDonald
    
Note: `'macdonald'.capitalize()` returns `'Macdonald'`

```python
def old_macdonald(name):
    return name[:3].capitalize() + name[3:].capitalize()
```

**Mi explicación (old_macdonald):** Parto la cadena en dos con slicing (los primeros 3 caracteres y el resto) y aplico capitalize() a cada mitad por separado, asi la primera Y la cuarta letra quedan en mayuscula.

```python
# Check
old_macdonald('macdonald')
```

#### REVERSA: Dada una oración, retornar la misma oración pero en reversa

    master_yoda('I am home') --> 'home am I'
    master_yoda('We are ready') --> 'ready are We'
    
Note: Revisar método .join()

    >>> "--".join(['a','b','c'])
    >>> 'a--b--c'

Resultado de .join():

    >>> " ".join(['Hello','world'])
    >>> "Hello world"

```python
def master_yoda(text):
    palabras = text.split()
    return ' '.join(palabras[::-1])
```

**Mi explicación (master_yoda):** Separo la oracion en palabras con split(), las invierto con el slicing [::-1] (el mismo stride que use en el cuaderno de cadenas) y las vuelvo a unir con join(' ').

```python
######
```

```python
# Revisión
master_yoda('I am home')
```

```python
# Revisión
master_yoda('We are ready')
```

#### PROBLEMA 33: 

Dada una lista de enteros, retornar True si el arreglo tinene un  3  seguido por un  3 .

    has_33([1, 3, 3]) → True
    has_33([1, 3, 1, 3]) → False
    has_33([3, 1, 3]) → False

```python
def has_33(nums):
    for i in range(len(nums) - 1):
        if nums[i] == 3 and nums[i + 1] == 3:
            return True
    return False
```

**Mi explicación (has_33):** Recorro la lista comparando cada elemento con el siguiente (nums[i] y nums[i+1]); apenas encuentro dos 3 seguidos retorno True de inmediato, y si el bucle termina sin encontrarlos retorno False.

```python
# Revisión
has_33([1, 3, 3])
```

```python
# Revisión
has_33([1, 3, 1, 3])
```

```python
# Revisión
has_33([3, 1, 3])
```

#### REPLICADOR: Dada una cadena, devuelve una cadena donde por cada carácter en el original hay tres caracteres REPETIDOS
    paper_doll('Hello') --> 'HHHeeellllllooo'
    paper_doll('Mississippi') --> 'MMMiiissssssiiippppppiii'

```python
def paper_doll(text):
    resultado = ''
    for letra in text:
        resultado += letra * 3
    return resultado
```

**Mi explicación (paper_doll):** Recorro cada caracter del texto y lo repito 3 veces con letra*3, acumulando todo en resultado; es basicamente aplicar la multiplicacion de cadenas que vimos en el cuaderno de cadenas pero caracter por caracter.

```python
# Check
paper_doll('Hello')
```

```python
# Check
paper_doll('Mississippi')
```

#### BLACKJACK: Dados tres enteros entre 1 y 11, si su suma es menor o igual a 21, devuelve su suma. Si su suma excede 21 *y* hay un once, reduce la suma total en 10. Finalmente, si la suma (incluso después del ajuste) excede 21, devuelve 'BUST'
    blackjack(5,6,7) --> 18
    blackjack(9,9,9) --> 'BUST'
    blackjack(9,9,11) --> 19

```python
def blackjack(a,b,c):
    total = a + b + c
    if total <= 21:
        return total
    elif total <= 31 and 11 in (a, b, c):
        return total - 10
    else:
        return 'BUST'
```

**Mi explicación (blackjack):** Primero calculo la suma total. Si ya es valida (<=21) la devuelvo tal cual. Si se pasa pero hay un 11 entre las cartas y restando 10 vuelve a ser valida, aplico ese ajuste (regla del As que vale 1 u 11). Si ni asi funciona, devuelvo 'BUST'.

```python
# Revisión
blackjack(5,6,7)
```

```python
# Revisión
blackjack(9,9,9)
```

```python
# Revisión
blackjack(9,9,11)
```
