## Comandos nuevos

- `print(x)` — muestra el valor de `x` en pantalla.

```python
print("Hola Mundo")
```

- `texto[i]` — indexación: accede al carácter en la posición `i` (0 es el primero).

```python
texto = "Python"
texto[0]
```

- `texto[-i]` — indexación negativa: cuenta las posiciones desde el final de la cadena.

```python
texto[-1]
```

- `len(texto)` — devuelve la cantidad de caracteres de la cadena.

```python
len(texto)
```

- `texto[a:b]` — slicing: extrae el fragmento desde el índice `a` hasta `b` (sin incluir `b`).

```python
texto[0:3]
```

- `texto[a:b:paso]` — stride: igual que el slicing pero saltando de a `paso` caracteres.

```python
texto[0:6:2]
```

- `texto1 + texto2` — concatena (une) dos cadenas en una nueva.

```python
"Hola" + " Mundo"
```

- `n * texto` — repite la cadena `n` veces.

```python
3 * "Ja"
```

- `\n`, `\t`, `\\` — secuencias de escape para salto de línea, tabulación y barra invertida literal.

```python
print("Linea1\nLinea2\tfin\\")
```

- `r"texto"` — cadena "cruda" (raw): ignora las secuencias de escape y muestra el texto tal cual.

```python
print(r"C:\nuevo\texto")
```

- `texto.upper()` — devuelve una copia de la cadena en mayúsculas.

```python
"hola".upper()
```

- `texto.replace(a, b)` — devuelve una copia de la cadena cambiando todas las apariciones de `a` por `b`.

```python
"hola mundo".replace("mundo", "Python")
```

- `texto.find(sub)` — devuelve el índice donde empieza `sub` dentro de la cadena, o `-1` si no la encuentra.

```python
"hola mundo".find("mundo")
```

## Solución del examen

**1.** ¿Cuál es el valor de la variable `A` tras ejecutar el siguiente código?

```python
A = "1"
A
```

**Explicación:** Es directo, `A` queda con el string `"1"` porque así se asignó, sin conversión de tipos de por medio.

**2.** ¿Cuál es el valor de la variable `B` tras ejecutar el siguiente código?

```python
B = "2"
B
```

**Explicación:** Igual que el caso anterior, `B` guarda el string `"2"` tal cual se escribió.

**3.** ¿Cuál es el valor de la variable `C` tras ejecutar `C = A + B`?

```python
C = A + B
C
```

**Explicación:** Como `A` y `B` son strings, el operador `+` concatena en vez de sumar numéricamente, por eso `C` queda como `"12"` y no como `3`.

**4.** En la variable `D`, usar slicing para imprimir solo los primeros tres elementos:

```python
D = "ABCDEFG"
print(D[:3])
```

**Explicación:** `D[:3]` toma desde el inicio hasta el índice 3 sin incluirlo, es decir las posiciones 0, 1 y 2.

**5.** Usar un stride de 2 para imprimir cada segundo elemento de la cadena `E`:

```python
E = 'clocrkr1e1c1t'
print(E[::2])
```

**Explicación:** El stride 2 salta de dos en dos empezando en el índice 0, así que toma un carácter sí y el siguiente no.

**6.** Imprimir una barra invertida:

```python
print("\\")
```

**Explicación:** Para imprimir una sola barra invertida hay que escaparla con otra barra invertida; si no, Python la interpreta como el inicio de una secuencia de escape.

**7.** Convertir a mayúsculas el contenido de la variable `F`:

```python
F = "You are wrong"
print(F.upper())
```

**Explicación:** `upper()` devuelve una copia de la cadena con todos los caracteres en mayúscula; no modifica `F` en el sitio.

**8.** En la variable `G`, encontrar el primer índice de la sub-cadena `snow` y reemplazar `Mary` por `Bob`:

```python
G = "Mary had a little lamb Little lamb, little lamb Mary had a little lamb \
Its fleece was white as snow And everywhere that Mary went Mary went, Mary went \
Everywhere that Mary went The lamb was sure to go"

print(G.find("snow"))
print(G.replace("Mary", "Bob"))
```

**Explicación:** `find` devuelve el índice donde empieza la primera aparición de `"snow"` dentro de `G`, y `replace` genera una cadena nueva cambiando todas las apariciones de `"Mary"` por `"Bob"` (`G` en sí no se modifica).
