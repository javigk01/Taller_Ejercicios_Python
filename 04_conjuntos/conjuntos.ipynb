## Descripción

Este cuaderno trabaja los conjuntos (`set`) en Python: cómo se crean con `{}` o convirtiendo una lista con `set()`, cómo Python elimina automáticamente los duplicados, y los métodos para modificarlos (`add`, `remove`) y verificar pertenencia (`in`). La segunda mitad se enfoca en las operaciones lógicas entre conjuntos: intersección, diferencia, unión, y las comprobaciones de subconjunto/superconjunto.

## Comandos nuevos

- `{a, b, c}` — crea un conjunto: colección **desordenada** de elementos únicos (sin duplicados).

```python
set1 = {"pop", "rock", "rock", "disco"}
set1
```

- `set(lista)` — convierte una lista (u otro iterable) en un conjunto, eliminando los duplicados.

```python
set([1, 2, 2, 1, 3])
```

- `conjunto.add(x)` — agrega un elemento al conjunto; si ya existe, no cambia nada.

```python
A = set(["Thriller", "AC/DC"])
A.add("NSYNC")
A
```

- `conjunto.remove(x)` — elimina un elemento del conjunto.

```python
A.remove("NSYNC")
A
```

- `x in conjunto` — verifica si un elemento pertenece al conjunto (devuelve `True` o `False`).

```python
"AC/DC" in A
```

- `A & B` — intersección: los elementos que están en ambos conjuntos a la vez.

```python
set1 = {"Thriller", "AC/DC", "Back in Black"}
set2 = {"AC/DC", "Back in Black", "The Wall"}
set1 & set2
```

- `A.difference(B)` — los elementos que están en `A` pero no en `B`.

```python
set1.difference(set2)
```

- `A.union(B)` — unión: todos los elementos de ambos conjuntos, sin repetir.

```python
set1.union(set2)
```

- `A.issuperset(B)` / `A.issubset(B)` — verifica si `A` contiene a `B`, o si `A` está contenido en `B`.

```python
set1.issuperset(set1 & set2), set2.issubset(set1.union(set2))
```

## Solución del examen

**1.** Transformar la lista `['rap', 'house', 'electronic music', 'rap']` en un conjunto:

```python
set(['rap', 'house', 'electronic music', 'rap'])
```

**Explicación:** Solo hacía falta envolver la lista con `set()`; el duplicado `'rap'` desaparece automáticamente en el resultado.

**2.** Dada la lista `A = [1, 2, 2, 1]` y el conjunto `B = set([1, 2, 2, 1])`, comparar `sum(A)` y `sum(B)`:

```python
A = [1, 2, 2, 1]
B = set([1, 2, 2, 1])
print("suma de A:", sum(A))
print("suma de B:", sum(B))
```

**Explicación:** `sum(A)` da `6` porque cuenta los duplicados de la lista, mientras que `sum(B)` da `3` porque el conjunto se quedó solo con los valores `1` y `2`. Es una buena forma de ver en la práctica que un conjunto sí cambia el resultado, no solo la forma de mostrarlo.

**3.** Crear un conjunto `album_set3` que sea la unión de `album_set1` y `album_set2`:

```python
album_set1 = set(["Thriller", "AC/DC", "Back in Black"])
album_set2 = set(["AC/DC", "Back in Black", "The Dark Side of the Moon"])

album_set3 = album_set1.union(album_set2)
album_set3
```

**Explicación:** `union()` junta todos los elementos de los dos conjuntos en uno solo, sin duplicar los que ya compartían (`AC/DC` y `Back in Black`).

**4.** Averiguar si `album_set1` es un subconjunto de `album_set3`:

```python
album_set1.issubset(album_set3)
```

**Explicación:** Da `True` porque `album_set3` se construyó justamente uniendo `album_set1` con otro conjunto, así que todos sus elementos ya estaban incluidos.
