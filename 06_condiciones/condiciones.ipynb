## Comandos nuevos

- `==`, `!=` — comparan si dos valores son iguales o distintos.

```python
a = 5
a == 6, a != 6
```

- `>`, `<`, `>=`, `<=` — comparan orden entre números, y también entre cadenas (por código ASCII).

```python
'B' > 'A', 3 <= 3
```

- `if condicion:` — ejecuta el bloque con sangría solo cuando la condición es verdadera.

```python
edad = 20
if edad > 18:
    print("puede entrar")
```

- `else:` — ejecuta un bloque alternativo cuando ninguna condición anterior fue verdadera.

```python
edad = 15
if edad > 18:
    print("puede entrar")
else:
    print("no puede entrar")
```

- `elif condicion:` — agrega una condición adicional a comprobar si la anterior fue falsa.

```python
edad = 18
if edad > 18:
    print("ve a ACDC")
elif edad == 18:
    print("ve a Pink Floyd")
else:
    print("ve a Meat Loaf")
```

- `and` — la condición combinada solo es verdadera si ambas partes lo son.

```python
(1980 > 1979) and (1980 < 1990)
```

- `or` — la condición combinada es verdadera si al menos una parte lo es.

```python
(1975 < 1980) or (1975 > 1989)
```

- `not` — invierte el valor de verdad de una condición.

```python
not (1983 == 1984)
```

## Solución del examen

**1.** Escribir una condición `if` que determine si un disco tiene una calificación mayor a 8, usando la calificación de "Back in Black" (8.5). Si es verdadera, imprimir `"This album is Amazing!"`:

```python
rating = 8.5
if rating > 8:
    print("This album is Amazing!")
```

**Explicación:** Como `8.5 > 8` es verdadero, el bloque con sangría se ejecuta e imprime el mensaje.

**2.** Escribir un `if`-`else`: si la calificación es mayor a 8 imprimir `"this album is amazing"`, si no imprimir `"this album is ok"`:

```python
rating = 8.5
if rating > 8:
    print("this album is amazing")
else:
    print("this album is ok")
```

**Explicación:** Con `rating = 8.5` la condición del `if` ya es verdadera, así que nunca se llega a evaluar el `else`.

**3.** Escribir un `if` que determine si un disco fue lanzado antes de 1980, o en 1991, o en 1993. Si la condición es verdadera, imprimir el año:

```python
album_year = 1979

if album_year < 1980 or album_year == 1991 or album_year == 1993:
    print(album_year)
```

**Explicación:** Encadené tres condiciones con `or`: basta con que se cumpla una sola para que se imprima el año; con `1979` se cumple la primera (`< 1980`).
