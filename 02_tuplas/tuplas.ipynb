## Descripción

Este cuaderno trabaja las tuplas en Python: cómo se crean, cómo se accede a sus elementos por índice (positivo y negativo), cómo se concatenan y se les aplica slicing, cómo se ordenan con `sorted` y cómo se accede a elementos cuando hay tuplas anidadas dentro de otras tuplas. La idea central que se repite es que una tupla es **inmutable**: cualquier operación que "modifica" una tupla en realidad genera una nueva.

## Comandos nuevos

- `(a, b, c)` — crea una tupla: colección ordenada e inmutable de valores.

```python
tupla = (10, "rock", 3.5)
tupla
```

- `type(x)` — devuelve el tipo de dato de `x`.

```python
type(tupla)
```

- `tupla[i]` — indexación: accede al elemento en la posición `i`.

```python
tupla[1]
```

- `tupla[-i]` — indexación negativa: cuenta las posiciones desde el final de la tupla.

```python
tupla[-1]
```

- `tupla1 + tupla2` — concatena dos tuplas en una nueva.

```python
tupla + ("pop", 2024)
```

- `tupla[a:b]` — slicing: extrae una sub-tupla entre los índices `a` y `b` (sin incluir `b`).

```python
(tupla + ("pop", 2024))[1:3]
```

- `len(tupla)` — devuelve la cantidad de elementos de la tupla.

```python
len(tupla)
```

- `sorted(tupla)` — devuelve una **lista** nueva con los elementos ordenados; la tupla original no cambia.

```python
sorted((0, 9, 6, 5, 10))
```

- `tupla[i][j]` — acceso anidado: entra a una tupla que está dentro de otra tupla.

```python
anidada = (1, 2, ("pop", "rock"))
anidada[2][0]
```

## Solución del examen

Se parte de la siguiente tupla de géneros musicales:

```python
genres_tuple = ("pop", "rock", "soul", "hard rock", "soft rock", \
                "R&B", "progressive rock", "disco")
genres_tuple
```

**1.** Encontrar la longitud de la tupla `genres_tuple`:

```python
len(genres_tuple)
```

**Explicación:** `len()` cuenta los elementos de la tupla directamente, sin importar el tipo de cada uno.

**2.** Acceder al elemento en el índice 3:

```python
genres_tuple[3]
```

**Explicación:** Contando desde 0, la posición 3 corresponde al cuarto elemento, `"hard rock"`.

**3.** Usar slicing para obtener los índices 3, 4 y 5:

```python
genres_tuple[3:6]
```

**Explicación:** Para incluir el índice 5 el límite superior del slice tiene que ser 6, ya que ese límite nunca se incluye.

**4.** Encontrar los primeros dos elementos de la tupla `genres_tuple`:

```python
genres_tuple[0:2]
```

**Explicación:** Un slice desde el inicio se puede escribir también como `genres_tuple[:2]`, pero dejé el `0` explícito para que quede claro el rango.

**5.** Encontrar el primer índice de `"disco"`:

```python
genres_tuple.index("disco")
```

**Explicación:** El método `index()` recorre la tupla y devuelve la posición de la primera coincidencia exacta del valor buscado.

**6.** Generar una lista ordenada a partir de la tupla `C_tuple = (-5, 1, -3)`:

```python
C_tuple = (-5, 1, -3)
C_list = sorted(C_tuple)
C_list
```

**Explicación:** `sorted()` no ordena la tupla en el sitio (no se puede, es inmutable); devuelve una lista nueva con los valores ordenados de menor a mayor.
