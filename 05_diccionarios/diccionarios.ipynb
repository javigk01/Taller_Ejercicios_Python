## Descripción

Este cuaderno trabaja los diccionarios en Python: colecciones de pares llave-valor donde, en vez de acceder por índice numérico como en una lista, se accede por una llave. Cubre cómo crearlos, cómo obtener un valor a partir de su llave, cómo listar todas las llaves (`keys()`) o todos los valores (`values()`), y cómo agregar, borrar y verificar la existencia de una entrada.

## Comandos nuevos

- `{llave: valor, ...}` — crea un diccionario; las llaves deben ser de un tipo inmutable (string, número o tupla).

```python
Dict = {"key1": 1, "key2": "2", (0, 1): 6}
Dict
```

- `dic[llave]` — accede al valor asociado a esa llave.

```python
Dict["key1"]
```

- `dic.keys()` — devuelve todas las llaves del diccionario.

```python
release_year_dict = {"Thriller": "1982", "Back in Black": "1980"}
release_year_dict.keys()
```

- `dic.values()` — devuelve todos los valores del diccionario.

```python
release_year_dict.values()
```

- `dic[nueva_llave] = valor` — agrega una entrada nueva, o actualiza el valor si la llave ya existía.

```python
release_year_dict["Graduation"] = "2007"
release_year_dict
```

- `del(dic[llave])` — elimina la entrada asociada a esa llave.

```python
del(release_year_dict["Graduation"])
release_year_dict
```

- `llave in dic` — verifica si una llave existe en el diccionario (devuelve `True` o `False`).

```python
"Thriller" in release_year_dict
```

## Solución del examen

Para las primeras dos preguntas se usa el siguiente diccionario:

```python
soundtrack_dic = {"The Bodyguard": "1992", "Saturday Night Fever": "1977"}
soundtrack_dic
```

**1a.** En el diccionario `soundtrack_dic`, ¿cuáles son las llaves?

```python
soundtrack_dic.keys()
```

**Explicación:** `keys()` devuelve las llaves del diccionario, en este caso los nombres de las dos películas.

**1b.** En el diccionario `soundtrack_dic`, ¿cuáles son los valores?

```python
soundtrack_dic.values()
```

**Explicación:** `values()` devuelve los valores asociados a esas llaves, en este caso los años de estreno.

**2a.** Crear un diccionario `album_sales_dict` con las ventas en millones de `Back in Black` (50), `The Bodyguard` (50) y `Thriller` (65):

```python
album_sales_dict = {"Back in Black": 50, "The Bodyguard": 50, "Thriller": 65}
album_sales_dict
```

**Explicación:** Cada disco queda como llave y su cifra de ventas como valor; que dos discos compartan el mismo valor (50) no genera ningún conflicto.

**2b.** Usar el diccionario para encontrar las ventas totales de `Thriller`:

```python
album_sales_dict["Thriller"]
```

**Explicación:** Basta con indexar el diccionario con la llave `"Thriller"` para obtener directamente su valor asociado.

**2c.** Encontrar los nombres de los discos dentro del diccionario con el método `keys`:

```python
album_sales_dict.keys()
```

**Explicación:** Reutilicé `keys()` igual que en la primera parte del examen, ahora sobre `album_sales_dict`.

**2d.** Encontrar los registros de ventas dentro del diccionario con el método `values`:

```python
album_sales_dict.values()
```

**Explicación:** Con `values()` se obtienen las cifras de ventas sin necesidad de saber a qué disco corresponde cada una.
