## Descripción

Este cuaderno trabaja los bucles en Python: `range()` para generar secuencias de números, el bucle `for` (recorriendo índices con `range` o directamente los elementos de una lista, incluido `enumerate` para tener índice y elemento a la vez), y el bucle `while`, que repite un bloque mientras una condición se mantenga verdadera. La diferencia clave que rescato es que `for` se usa cuando se sabe de antemano cuántas veces se va a repetir (o sobre qué se va a iterar), mientras que `while` se usa cuando la repetición depende de una condición que puede cambiar en cualquier iteración.

## Comandos nuevos

- `range(n)` — genera una secuencia de números desde `0` hasta `n-1`.

```python
list(range(3))
```

- `range(a, b)` — genera una secuencia de números desde `a` hasta `b-1`.

```python
list(range(2, 8))
```

- `for i in range(n):` — repite un bloque de código `n` veces, usando `i` como contador.

```python
for i in range(3):
    print("vuelta", i)
```

- `for elemento in lista:` — recorre directamente los elementos de una lista, sin necesidad de índices.

```python
generos = ["rock", "pop", "soul"]
for genero in generos:
    print(genero)
```

- `enumerate(lista)` — entrega a la vez el índice y el elemento en cada iteración del `for`.

```python
colores = ["rojo", "verde", "azul"]
for i, color in enumerate(colores):
    print(i, color)
```

- `while condicion:` — repite un bloque de código mientras la condición se mantenga verdadera.

```python
i = 0
while i < 3:
    print(i)
    i = i + 1
```

## Solución del examen

**1.** Escribir un bucle `for` que imprima todos los elementos entre -5 y 5 usando `range`:

```python
for i in range(-5, 6):
    print(i)
```

**Explicación:** El límite superior de `range` no se incluye, así que para llegar hasta el 5 hay que pedir `range(-5, 6)`.

**2.** Imprimir los elementos de `Genres = ['rock', 'R&B', 'Soundtrack', 'R&B', 'soul', 'pop']` siguiendo las convenciones del lenguaje:

```python
Genres = ['rock', 'R&B', 'Soundtrack', 'R&B', 'soul', 'pop']
for genre in Genres:
    print(genre)
```

**Explicación:** Al no necesitar el índice, iterar directamente sobre la lista (`for genre in Genres`) es más limpio que recorrerla con `range(len(Genres))`.

**3.** Escribir un bucle `for` que imprima la lista `squares = ['red', 'yellow', 'green', 'purple', 'blue']`:

```python
squares = ['red', 'yellow', 'green', 'purple', 'blue']
for square in squares:
    print(square)
```

**Explicación:** Mismo patrón que el ejercicio anterior, iterando directo sobre los elementos de la lista.

**4.** Escribir un bucle `while` que muestre las calificaciones de `PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]` y se detenga apenas encuentre una calificación menor a 6:

```python
PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]
i = 1
Rating = PlayListRatings[0]
while Rating >= 6:
    print(Rating)
    Rating = PlayListRatings[i]
    i = i + 1
```

**Explicación:** Uso `while` en vez de `for` porque no sé de antemano cuántos elementos voy a recorrer; la condición `Rating >= 6` es la que decide cuándo parar, y se corta apenas aparece el 5.

**5.** Escribir un bucle `while` que copie las cadenas `'orange'` de `squares` a `new_squares`, deteniéndose en el primer valor que no sea `'orange'`:

```python
squares = ['orange', 'orange', 'purple', 'blue ', 'orange']
new_squares = []
i = 0
while squares[i] == 'orange':
    new_squares.append(squares[i])
    i = i + 1
new_squares
```

**Explicación:** El bucle revisa `squares[i]` antes de agregarlo; en cuanto encuentra `'purple'` (que no es `'orange'`) la condición se vuelve falsa y el ciclo termina, dejando el `'orange'` final de la lista sin copiar.
