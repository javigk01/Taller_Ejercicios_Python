## Descripción

Este cuaderno trabaja las listas en Python: cómo se crean, cómo se accede a sus elementos por índice, cómo se les aplica slicing, y los métodos para modificarlas (`extend`, `append`, asignación por índice, `del`). También cubre `split` para convertir una cadena en lista, y un punto que me pareció clave: la diferencia entre copiar una lista por **referencia** (`B = A`, ambas variables apuntan a la misma lista) y **clonarla** (`B = A[:]`, se crea una lista independiente). A diferencia de las tuplas, las listas sí son mutables.

## Comandos nuevos

- `[a, b, c]` — crea una lista: colección ordenada y **mutable** de valores.

```python
L = ["Michael Jackson", 10.1, 1982]
L
```

- `lista[i]` / `lista[-i]` — indexación positiva y negativa, igual que en cadenas y tuplas.

```python
L[0], L[-1]
```

- `lista[a:b]` — slicing: extrae una sub-lista entre los índices `a` y `b` (sin incluir `b`).

```python
L = ["Michael Jackson", 10.1, 1982, "MJ", 1]
L[3:5]
```

- `lista.extend([...])` — agrega varios elementos al final de la lista, modificándola en el sitio.

```python
L = ["Michael Jackson", 10.2]
L.extend(["pop", 10])
L
```

- `lista.append(x)` — agrega un solo elemento al final (si `x` es una lista, se agrega como un elemento anidado, no se combinan).

```python
L.append(["a", "b"])
L
```

- `lista[i] = valor` — cambia el elemento en la posición `i` (las listas sí se pueden modificar así, a diferencia de las tuplas).

```python
A = ["disco", 10, 1.2]
A[0] = "hard rock"
A
```

- `del(lista[i])` — elimina el elemento en la posición `i`.

```python
del(A[0])
A
```

- `"texto".split(sep)` — separa una cadena en una lista de partes usando `sep` como separador (por defecto, espacios).

```python
"A,B,C,D".split(",")
```

- `B = A` vs `B = A[:]` — la primera copia por **referencia** (A y B apuntan a la misma lista); la segunda **clona** la lista (B queda independiente).

```python
A = ["hard rock", 10, 1.2]
B = A[:]
A[0] = "banana"
A, B
```

## Solución del examen

**1.** Crear una lista `a_list` con los elementos `1`, `"hello"`, `[1, 2, 3]` y `True`:

```python
a_list = [1, 'hello', [1, 2, 3], True]
a_list
```

**Explicación:** Una lista admite tipos distintos mezclados en el mismo objeto, incluida otra lista anidada como tercer elemento.

**2.** Encontrar el valor almacenado en el índice 1 de `a_list`:

```python
a_list[1]
```

**Explicación:** El índice 1 corresponde al segundo elemento, que es el string `'hello'`.

**3.** Obtener los elementos almacenados en los índices 1, 2 y 3 de `a_list`:

```python
a_list[1:4]
```

**Explicación:** Para incluir el índice 3 en el slice el límite superior tiene que ser 4, ya que ese límite no se incluye.

**4.** Concatenar las listas `A = [1, 'a']` y `B = [2, 1, 'd']`:

```python
A = [1, 'a']
B = [2, 1, 'd']
A + B
```

**Explicación:** Igual que con las cadenas y las tuplas, el operador `+` entre listas concatena y devuelve una lista nueva con los elementos de ambas en orden.
