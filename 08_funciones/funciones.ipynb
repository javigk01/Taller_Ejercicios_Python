## Comandos nuevos

- `def nombre(params):` ... `return valor` — define una función y le indica qué devolver al llamarla.

```python
def sumar_uno(a):
    b = a + 1
    return b

sumar_uno(5)
```

- `help(funcion)` — muestra la documentación (docstring) asociada a una función.

```python
help(sumar_uno)
```

- `def f(x=valor):` — define un parámetro con un valor por defecto, que se usa si no se pasa nada en la llamada.

```python
def es_buena_calificacion(rating=4):
    if rating < 7:
        return "mala"
    else:
        return "buena"

es_buena_calificacion(), es_buena_calificacion(10)
```

- Variable **local** vs **global** — una variable creada dentro de una función solo existe ahí; una creada afuera (global) se puede leer desde cualquier función.

```python
artista = "AC/DC"

def mostrar_artista():
    print(artista, "es una banda")

mostrar_artista()
```

- `global variable` — dentro de una función, permite crear o modificar una variable del ámbito global (no solo leerla).

```python
contador = 0

def incrementar():
    global contador
    contador = contador + 1

incrementar()
incrementar()
contador
```

- Combinar `for`/`if` dentro de una función — el cuerpo de una función puede contener cualquier estructura de control.

```python
def clasificar_anios(anios):
    for anio in anios:
        if anio > 1980:
            print(anio, "-> moderno")
        else:
            print(anio, "-> antiguo")

clasificar_anios([1973, 1982, 1990])
```

## Solución del examen

**1.** Crear una función que divida la primera entrada por la segunda:

```python
def dividir(a, b):
    return a / b

dividir(10, 2)
```

**Explicación:** La función solo necesita devolver `a / b`; no hace falta ninguna condición porque el enunciado no pide manejar la división por cero.

Para las siguientes dos preguntas se usa esta función, definida en el cuaderno original:

```python
def con(a, b):
    return a + b
```

**2.** ¿Se puede usar la función `con` para sumar números enteros o cadenas?

```python
con(2, 2), con("a", "b")
```

**Explicación:** Sí, con ambos tipos funciona porque el operador `+` está definido tanto para sumar números como para concatenar cadenas; `con` no le pone ninguna restricción al tipo de dato.

**3.** ¿Se puede usar la función `con` para concatenar una lista o una tupla?

```python
con(['a', 1], ['b', 1])
```

**Explicación:** También funciona, porque `+` entre listas concatena igual que entre cadenas; `con` es genérica en ese sentido, sirve para cualquier tipo que soporte el operador `+`.
